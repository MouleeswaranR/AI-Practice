```python
#how to tokenize large dataset faster
import time

text= ["This is a simpe sentence for tokenization. "]*100000
```

```python
#Creating own tokenizer

from tokenizers import Tokenizer, models, trainers, pre_tokenizers 
corpus=["I love transformers!","Tokenizers are amazing.","Build your own Tokenizer.\n"]
```

```python
#Create a new tokenizer with byte-pair encoding model
tokenizer=Tokenizer(models.BPE())

#pre tokenizer for the new tokenizer
tokenizer.pre_tokenizer=pre_tokenizers.Whitespace()

#Byte Pair Encoding trainer for new tokenizer with its vocab size
trainer=trainers.BpeTrainer(vocab_size=1000)

#train the new tokenizer
tokenizer.train_from_iterator(corpus,trainer)
```

```python
#save the new tokenizer
tokenizer.save("custom-tokenizer.json")
```

```python
#cosine similarity

from torch.nn.functional import cosine_similarity
```

```python
sentence1="Dhoni is my favorite cricketer"
sentence2="Vijay is my favorite actor"
```

```python
input1=tokenizer(sentence1,return_tensors="pt")
input2=tokenizer(sentence2,return_tensors="pt")
```

```python
with torch.no_grad():
    output1=model(**input1).last_hidden_state[0].mean(dim=0)
    output2=model(**input2).last_hidden_state[0].mean(dim=0)
```

```python
similarity=cosine_similarity(output1,output2,dim=0)
print(f"Similarity: {similarity.item():.4f}")
```

```python
import evaluate
```

```python
accuracy = evaluate.load("accuracy")
```

```python
result=accuracy.compute(predictions=[0,1,1,0],references=[0,1,0,0])
```

```python
print(result)
```

```python
def compute_metrics(eval_pred):
    logits,labels=eval_pred
    predictions=logits.argmax(axis=-1)
    return accuracy.compute(predictions=predictions,references=labels)
```

```python
from transformers import AutoTokenizer,AutoModelForSequenceClassification
from transformers import Trainer, TrainingArguments
```

```python
#load datsets
from datasets import load_dataset
train_dataset=load_dataset("imdb",split="train[:100]")
test_dataset=load_dataset("imdb",split="test[:100]")
```

```python
model=AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased",num_labels=2)
tokenizer=AutoTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
def tokenize(example):
    return tokenizer(example['text'],truncation=True,padding="max_length")
```

```python
trained_encoded=train_dataset.map(tokenize,batched=True)
test_encoded=test_dataset.map(tokenize,batched=True)
```

```python
trained_encoded.set_format("torch",columns=["input_ids","attention_mask","label"])
test_encoded.set_format("torch",colums=["input_ids","attention_mask","label"])
```

```python
%pip install accelerate>=1.1.0
```

```python
args=TrainingArguments(output_dir="eval-check",per_device_eval_batch_size=32,report_to="none")
```

```python
trainer=Trainer(model=model,args=args,compute_metrics=compute_metrics)
```

```python
metrics=trainer.evaluate(eval_dataset=test_encoded)
```

```python
print(metrics)
```

```python
#bleu(Bilingual Evaluation Understudy) score for translation/text-generation tasks evaluation(precision)

from evaluate import load

bleu=load("bleu")

predictions=["The cat is sitting on the mat"]

references=[["The cat sat on the mat"]]

result=bleu.compute(predictions=predictions,references=references)
print("BLEU Score: ",result)
```

```python
#rouge((Recall-Oriented Understudy for Gisting Evaluation) score for summarization tasks (recall)

rouge=load("rouge")

predictions=["The cat is sitting on the mat"]

references=[["The cat sat on the mat"]]

result=rouge.compute(predictions=predictions,references=references)

print("rouge score:",result)
```
