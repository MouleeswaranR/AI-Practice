```python
from datasets import load_dataset,Dataset
```

```python
dataset=load_dataset("roneneldan/TinyStories",split="train")
```

```python
print(dataset)
```

```python
#load custom data from pdf
import fitz

def extract_text_from_pdf(pdf_path):
    text_block=[]

    with fitz.open(pdf_path) as doc:
        for page in doc:
            text=page.get_text("text").strip()
            if text:
                text_block.append(text)
        
    return text_block


```

```python
pdf_texts=extract_text_from_pdf(r"C:\Users\INMOR14\OneDrive - ABB\Documents\finetuning\The Importance of Continuous Learning in a Changing World.pdf")
```

```python
print(pdf_texts)
```

```python
import re
def split_paragraphs(pages):
    paragraphs=[]

    for page_text in pages:
        chunks=re.split(r'\n\s*\n',page_text)

        for chunk in chunks:
            clean=chunk.strip()
            if len(clean)>30:
                paragraphs.append(chunk)

    return paragraphs
```

```python
paragraphs=split_paragraphs(pdf_texts)
```

```python
paragraphs
```

```python
data=[{"text":p} for p in paragraphs]
```

```python
data
```

```python
dataset=Dataset.from_list(data)
```

```python
dataset
```

```python
model_name="TinyLlama/TinyLlama-1.1B-intermediate-step-1431k-3T"
```

```python
from transformers import AutoTokenizer,AutoModelForCausalLM,Trainer,TrainingArguments,DataCollatorForLanguageModeling
```

```python
tokenizer=AutoTokenizer.from_pretrained(model_name)
```

```python
if tokenizer.pad_token is None:
    tokenizer.pad_token=tokenizer.eos_token
```

```python
def tokenize_fn(examples):
    tokens=tokenizer(examples['text'],truncation=True,padding="max_length",max_length=512)
    tokens["labels"]=tokens["input_ids"].copy()
    return tokens
```

```python
tokenized=dataset.map(tokenize_fn,batched=True,remove_columns=["text"])
```

```python
model=AutoModelForCausalLM.from_pretrained(model_name)
```

```python
training_args=TrainingArguments(
    output_dir="./instruction-finetuning",
    num_train_epochs=2,
    per_device_train_batch_size=1,
    save_steps=500,
    save_total_limit=2,
    learning_rate=2e-5,
    logging_steps=50,
    fp16=True,
    report_to="none"
)
```

```python
trainer=Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized
)
```

```python
from peft import LoraConfig,get_peft_model,TaskType
```

```python
tokenized=dataset.map(tokenize_fn,batched=True)
```

```python
model = "TinyLlama/TinyLlama-1.1B-intermediate-step-1431k-3T"
```

```python
from transformers import BitsAndBytesConfig

bnb_config=BitsAndBytesConfig(
    load_in_8bit=True
)

model = AutoModelForCausalLM.from_pretrained(
    model,
    quantization_config=bnb_config,
    device_map="auto"
)
```

```python
lora_config=LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=8,
    lora_alpha=16,
    target_modules=["q_proj","v_proj"],
    lora_dropout=0.05,
    bias="none"
)
```

```python
non_inst_model_lora=get_peft_model(model,lora_config)
```

```python
training_args=TrainingArguments(
    output_dir='./lora_trained_model',
    num_train_epochs=5,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    fp16=True,
    learning_rate=2e-4,
    logging_steps=20,
    save_total_limit=1,
    report_to="none"
)
```

```python
trainer=Trainer(
    model=non_inst_model_lora,
    args=training_args,
    train_dataset=tokenized
)
```

```python
trainer.train()
```

```python
model_path="./lora_trained_model/checkpoint-5"
```

```python
model = AutoModelForCausalLM.from_pretrained(model_path, device_map="auto")
```

```python
prompt="Continuous learning is not limited to formal education. It includes"
```

```python
inputs=tokenizer(prompt,return_tensors="pt")
```

```python
outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    temperature=0.8,
    top_p=0.9,
    do_sample=True,
    repetition_penalty=1.1
)
```
